## Estadística Inferencial, formulación y prueba de hipótesis

### Conceptos básicos de la Estadística Inferencial:



#### Diferencia entre estadística descriptiva e inferencial:

- Población: Conjunto total de elementos que se estudian
- Muestra: Subconjunto representativo de la población
- Muestreo Aleatorio: Método de selección de la muestra que asegura que cada individuo tiene la misma probabilidad de ser elegido, reduciendo sesgos.

#### Parámetros y estimadores:
- Parámetro poblacional: Valor numérico que describe una característica de la población
- Estadístico muestral: Valor numérico calculado a partir de la muestra, usado para calcular el parámetro poblacional

#### Ejemplos:

##### 1. Población

La población es el conjunto **completo** de unidades estadísticas
sobre las que se quiere hacer una afirmación.
Es el universo de referencia del estudio.

> La definición de la población depende del objeto de investigación:
> no es lo mismo estudiar "los argentinos" que
> "los legisladores nacionales" o "los partidos provinciales".

| Pregunta de investigación |
|:-------------------------- |
| ¿Cuántos argentinos aprueban la gestión presidencial? |
| ¿Qué porcentaje de legisladores nacionales apoyó la reforma previsional? |
| ¿Cuántos municipios bonaerenses tienen población de mas de 100.000 habitantes? |
| ¿Qué proporción de partidos provinciales sobrevivió más de dos elecciones? |

> **Nota clave:** cuando la población es pequeña y accesible
> (como los 257 diputados), es posible relevarla completamente
> y obtener el parámetro directamente, sin necesidad de muestreo.
> Cuando es grande (el padrón electoral), el muestreo es indispensable.

##### 2. Muestra

La muestra es un **subconjunto** de la población,
seleccionado de forma tal que sea representativo
de las características del universo completo.

$$n < N$$

El tamaño de la muestra (n) es siempre menor que
el tamaño de la población (N).

**Estudio:** aprobación presidencial en Argentina.

- **Población (N):** ~33 millones de ciudadanos habilitados
- **Muestra (n):** 1.200 ciudadanos seleccionados aleatoriamente

| ID | Provincia    | Edad | Sexo | Aprueba gestión | Confianza inst. (0–10) |
|:--:|:-------------|:----:|:----:|:---------------:|:----------------------:|
| 1  | Buenos Aires | 34   | F    | Sí              | 6                      |
| 2  | Córdoba      | 52   | M    | No              | 3                      |
| 3  | Santa Fe     | 27   | F    | No              | 2                      |
| 4  | Mendoza      | 45   | M    | Sí              | 7                      |
| 5  | Tucumán      | 61   | F    | No              | 4                      |
| …  | …            | …    | …    | …               | …                      |
| 1200 | Neuquén  | 38   | M    | Sí              | 5                      |

> La muestra no es cualquier subconjunto: debe ser
> seleccionada con un procedimiento que garantice
> representatividad. De lo contrario, las inferencias
> sobre la población serán sesgadas.

##### 3. Muestreo aleatorio

El muestreo aleatorio es el procedimiento por el cual
cada unidad de la población tiene una **probabilidad
conocida y no nula** de ser seleccionada.
Es la condición que permite hacer inferencia válida.

#### Tipos principales y ejemplos políticos

##### Muestreo aleatorio simple

Cada ciudadano tiene exactamente la misma probabilidad
de ser seleccionado.

$$P(\text{selección}) = \frac{n}{N} = \frac{1200}{33.000.000} \approx 0{,}0036\%$$

**Ejemplo:** Se extrae al azar un listado de 1.200 DNI
del padrón electoral nacional para encuestar sobre
aprobación presidencial.

##### Muestreo estratificado

La población se divide en subgrupos (estratos) según
una variable relevante, y se selecciona aleatoriamente
dentro de cada estrato. Garantiza representación proporcional.

**Ejemplo:** Se divide el padrón en 24 distritos electorales
y se seleccionan encuestados en proporción al tamaño
del electorado de cada provincia.

| Provincia    | Electorado | % del total | n asignado |
|:-------------|:----------:|:-----------:|:----------:|
| Buenos Aires | 12.000.000 | 36,4%       | 437        |
| Córdoba      | 3.800.000  | 11,5%       | 138        |
| Santa Fe     | 3.500.000  | 10,6%       | 127        |
| Resto        | 13.700.000 | 41,5%       | 498        |
| **Total**    |**33.000.000**|**100%**   | **1.200**  |

##### Muestreo por conglomerados

Se seleccionan aleatoriamente grupos (conglomerados)
y se relevan todas las unidades dentro de ellos.

**Ejemplo:** Se sortean aleatoriamente 60 municipios
del país y se encuesta a todos los concejales de
esos municipios para estudiar representación de género
en los cuerpos legislativos locales.

##### Muestreo sistemático

Se elige un punto de inicio al azar y luego se selecciona
cada k-ésima unidad del listado.

$$k = \frac{N}{n} = \frac{33.000.000}{1.200} = 27.500$$

**Ejemplo:** Del padrón ordenado por DNI, se elige
aleatoriamente un número entre 1 y 27.500 como punto
de inicio y luego se selecciona cada 27.500 registros.

##### 4. Parámetro poblacional

El parámetro es una medida numérica que describe
una característica de la **población completa**.
Es un valor **fijo pero generalmente desconocido**,
porque relevarlo implicaría censar a toda la población.

Se denota con letras griegas:

| Parámetro | Símbolo | Descripción |
|:----------|:-------:|:------------|
| Proporción poblacional | π | Fracción de la población con cierta característica |
| Media poblacional | μ | Promedio de la población |
| Varianza poblacional | σ² | Dispersión en la población |
| Desvío estándar poblacional | σ | Raíz de la varianza poblacional |

**Ejemplo 1 — proporción:**
El verdadero porcentaje de ciudadanos argentinos que aprueba
la gestión del presidente en este momento.
Nadie lo conoce con exactitud: es el parámetro π.

**Ejemplo 2 — media:**
El verdadero promedio de confianza institucional
(medido en escala 0–10) de todos los argentinos mayores de 18 años.
Es el parámetro μ.

**Ejemplo 3 — varianza:**
La verdadera dispersión del porcentaje de votos obtenido
por un partido en todos los municipios del país.
Es el parámetro σ².

> Los parámetros existen en la realidad pero son inobservables
> en poblaciones grandes. La estadística inferencial permite
> estimarlos a partir de los estadísticos muestrales.

## 5. Estadístico muestral

El estadístico es una medida numérica calculada
a partir de los datos de la **muestra**.
Es observable, calculable, y se usa para **estimar**
el parámetro poblacional correspondiente.

Se denota con letras latinas o con "sombrero":

| Estadístico | Símbolo | Estima al parámetro |
|:------------|:-------:|:-------------------:|
| Proporción muestral | p̂ | π |
| Media muestral | x̄ | μ |
| Varianza muestral | s² | σ² |
| Desvío estándar muestral | s | σ |

##### Encuesta de aprobación presidencial

De los 1.200 encuestados:

- 624 aprueban la gestión del presidente
- 576 no aprueban

**Proporción muestral (estadístico):**

$$\hat{p} = \frac{624}{1200} = 0{,}52 \quad \Rightarrow \quad 52\%$$

**Media muestral de confianza institucional (estadístico):**

$$\bar{x} = \frac{\sum_{i=1}^{1200} x_i}{1200} = 4{,}3 \text{ puntos sobre 10}$$

**Desvío estándar muestral (estadístico):**

$$s = 2{,}1 \text{ puntos}$$

Estos tres valores son estadísticos: los calculamos
a partir de la muestra y los usamos para estimar
los parámetros reales de los 33 millones de argentinos.

#### Resumen integrador

| Concepto | Símbolo | ¿Observable? | Ejemplo |
|:---------|:-------:|:------------:|:--------|
| Población (N) | N | No completamente | ~33 millones de argentinos |
| Muestra (n) | n | Sí | 1.200 encuestados |
| Muestreo aleatorio | — | Sí (el procedimiento) | Sorteo del padrón electoral |
| Parámetro poblacional | π, μ, σ | No | Verdadero % de aprobación presidencial |
| Estadístico muestral | p̂, x̄, s | Sí | 52% de aprobación en la muestra |

La inferencia estadística es el puente entre
el estadístico (lo que medimos) y el parámetro
(lo que queremos conocer). Nunca alcanzamos el parámetro
con certeza absoluta, pero podemos estimarlo con
un margen de error calculable y una confianza conocida.

## Formulación de Hipótesis


Una **hipótesis estadística** es una afirmación sobre una característica de la población que se somete a prueba mediante su análisis de una muestra. En estadística inferencial toda prueba parte de dos hipótesis contrapuestas sobre un parámetro poblacional desconocido.
El objetivo es decidir, a partir de los datos de la muestra, cuál de las dos es más compatible con la evidencia.
Existen distintos tipos de hipótesis:

- **Hipótesis nula:** $$H_0$$ Representa la afirmación que se somete a prueba y que se supone cierta hasta que se demuestre lo contrario. En términos matemáticos suele formularse con una igualdad.
**Siempre se expresa con igualdad (=)**  porque afirma que el parámetro tiene un valor específico o que dos grupos no difieren entre sí.

Símbolos posibles en H₀: $=$ $\leq$ $\geq$

- **Hipótesis alternativa:** $$H_1$$ Es la hipótesis que plantea una diferencia significativa respecto de la hipótesis nula. Es la afirmación que el investigador busca demostrar y generalmente se expresa en desigualdad.
**Siempre se expresa con desigualdad** porque afirma que el parámetro es distinto, mayor o menor que el valor planteado en H₀.

Símbolos posibles en H₁: $\neq$ $>$ $<$


### Regla práctica para la clase. Ejemplos:

| | H₀ | H₁ |
|:--|:--:|:--:|
| ¿Qué afirma? | No hay efecto / no hay diferencia | Hay efecto / hay diferencia |
| ¿Qué símbolo usa? | = (o ≤ o ≥) | ≠ (o > o <) |
| ¿Se puede "probar"? | No: solo se rechaza o no se rechaza | Es lo que se intenta sostener |
| Analogía jurídica | Presunción de inocencia | Cargo que debe probarse |

> La analogía jurídica es útil para entender: así como un acusado se presume inocente hasta que se pruebe lo contrario, H₀ se presume verdadera hasta que los datos aporten evidencia suficiente para rechazarla.

## Las colas de la distribución

El concepto de "cola" se refiere a las zonas extremas de la distribución muestral del estadístico de prueba. Si el valor calculado cae en la cola (región de rechazo), se rechaza H₀.

### 2. Hipótesis unilaterales (prueba de una cola)

Una hipótesis es **unilateral** cuando H₁ especifica **la dirección** del efecto: mayor que o menor que. Se usa cuando la teoría o evidencia previa indica
en qué sentido se espera el cambio.
La región de rechazo se concentra en **una sola cola** de la distribución: solo la derecha o solo la izquierda.

### 3. Hipótesis bilaterales (prueba de dos colas)

Una hipótesis es **bilateral** cuando H₁ plantea que el parámetro es **distinto** del valor de referencia, sin especificar en qué dirección. Se usa cuando no hay teoría previa que indique si el efecto será positivo o negativo.

$$H_0: \pi = \pi_0 \qquad H_1: \pi \neq \pi_0$$

La región de rechazo se divide en **dos colas** de la distribución: una en el extremo inferior y otra en el superior.

## 4. Resumen comparativo

| | Bilateral | Cola derecha | Cola izquierda |
|:--|:--:|:--:|:--:|
| H₀ | π = π₀ | π ≤ π₀ | π ≥ π₀ |
| H₁ | π ≠ π₀ | π > π₀ | π < π₀ |
| Colas | 2 | 1 (derecha) | 1 (izquierda) |
| Región de rechazo | Ambos extremos | Extremo derecho | Extremo izquierdo |
| z* (α = 0,05) | ±1,96 | +1,645 | −1,645 |
| ¿Cuándo usarla? | Sin dirección teórica | Se espera aumento | Se espera caída |
| Ejemplo político | ¿Aprobación ≠ 50%? | ¿Aprobación > 50% en Patagonia? | ¿Confianza < 4,5 histórico? |

> **Regla práctica:** si la pregunta de investigación
> usa "¿es diferente?", la prueba es bilateral.
> Si usa "¿es mayor?" o "¿es menor?", la prueba es unilateral.
> La dirección la da siempre la teoría, nunca los datos
> (usar los datos para elegir la cola es una mala práctica).

## Nivel de significancia, nivel de confianza e intervalos

### 1. Nivel de significancia (α)

El nivel de significancia es la probabilidad máxima que el investigador acepta de **cometer un error al rechazar H₀ cuando en realidad es verdadera**.
Se fija *antes* de recolectar los datos.

$$\alpha = P(\text{rechazar } H_0 \mid H_0 \text{ es verdadera})$$

Es el umbral de tolerancia al error. Cuanto más chico, más exigente es la prueba.

### Valores convencionales en Ciencia Política

| α | Nivel de exigencia | Interpretación |
|:--:|:--:|:--|
| 0,10 | Bajo | Se acepta hasta un 10% de probabilidad de error |
| 0,05 | Estándar | El más usado en ciencias sociales |
| 0,01 | Alto | Se exige evidencia muy fuerte para rechazar H₀ |

### Ejemplo

**Pregunta:** ¿La aprobación presidencial es distinta del 50%?

$$H_0: \pi = 0{,}50 \qquad H_1: \pi \neq 0{,}50$$

Con **α = 0,05** el investigador acepta que, si rechaza H₀,
existe un 5% de probabilidad de estar equivocado
(es decir, de rechazar algo que en realidad era verdadero).

> En Ciencia Política se usa casi siempre α = 0,05
> porque las variables sociales tienen alta variabilidad
> natural y α = 0,01 resultaría demasiado restrictivo.

## 2. Nivel de confianza (1 − α)

El nivel de confianza es el complemento del nivel de significancia. Expresa la probabilidad de que el intervalo construido **contenga al verdadero parámetro**.

$$\text{Nivel de confianza} = 1 - \alpha$$

| α | Nivel de confianza | z crítico |
|:--:|:--:|:--:|
| 0,10 | 90% | 1,645 |
| 0,05 | 95% | 1,960 |
| 0,01 | 99% | 2,576 |

### Ejemplo

Con α = 0,05 → nivel de confianza = **95%**.

Esto significa: si se construyeran 100 intervalos a partir de 100 muestras distintas del padrón electoral, aproximadamente 95 de ellos contendrían el verdadero porcentaje de aprobación presidencial de toda la población.

## 3. Intervalo de confianza

Un intervalo de confianza es un rango de valores construido a partir del estadístico muestral, dentro del cual se estima que se encuentra el parámetro poblacional con un nivel de confianza dado. Un intervalo de Confianza del 95% implica que, si repitiesemos el experimento muchas veces y calculáramos el intervalo de confianza para *cada muestra*, aproximadamente el 95% de esos intervalos contendrían el valor real del parámetro. En otras palabras, hay un 95% de probabilidad de que el valor verdadero del parámetro se encuentre *dentro del intervalo estimado*

$$IC = \text{estadístico} \pm \text{margen de error}$$

## 4. Margen de error

El margen de error (E) es la mitad del ancho del intervalo. Indica cuánto puede alejarse el estadístico del parámetro real con el nivel de confianza elegido. Los márgenes de error suelen reportarse en encuestas e investigaciones y son muy importantes a la hora de extraer conclusiones de los parámetros presentados.

$$E = z_{\alpha/2} \cdot \sigma_{\hat{p}} \quad \text{(para proporciones)}$$

$$E = z_{\alpha/2} \cdot \frac{s}{\sqrt{n}} \quad \text{(para medias)}$$

### Ejemplo: aprobación presidencial

De una muestra de n = 1.200 ciudadanos argentinos, 624 aprueban la gestión del presidente.

$$\hat{p} = \frac{624}{1200} = 0{,}52$$

Con nivel de confianza del 95% → $z_{\alpha/2} = 1{,}96$

**Error estándar:**

$$\sigma_{\hat{p}} = \sqrt{\frac{0{,}52 \times 0{,}48}{1200}} = \sqrt{\frac{0{,}2496}{1200}} = \sqrt{0{,}000208} \approx 0{,}0144$$

**Margen de error:**

$$E = 1{,}96 \times 0{,}0144 \approx 0{,}0282 \approx 2{,}82\%$$

**Intervalo de confianza:**

$$IC(\pi) = 0{,}52 \pm 0{,}0282$$

$$IC(\pi) = [0{,}4918 \; ; \; 0{,}5482]$$

$$IC(\pi) = [49{,}18\% \; ; \; 54{,}82\%]$$

**Interpretación:** Con un 95% de confianza, el verdadero porcentaje de aprobación presidencial entre todos los argentinos se encuentra entre el **49,18% y el 54,82%**.

### Efecto del nivel de confianza sobre el ancho del intervalo

Usando los mismos datos (n=1.200, p̂=0,52):

| Nivel de confianza | z | Margen de error | Intervalo |
|:--:|:--:|:--:|:--|
| 90% | 1,645 | ±2,37% | [49,63% ; 54,37%] |
| 95% | 1,960 | ±2,82% | [49,18% ; 54,82%] |
| 99% | 2,576 | ±3,71% | [48,29% ; 55,71%] |

A mayor nivel de confianza, mayor margen de error y más ancho el intervalo. Ganar certeza tiene un costo: menos precisión.

### Efecto del tamaño muestral sobre el margen de error

Con p̂ = 0,52 y nivel de confianza del 95%:

| n | Margen de error |
|:--:|:--:|
| 200  | ±6,92% |
| 400  | ±4,89% |
| 800  | ±3,46% |
| 1.200 | ±2,82% |
| 2.400 | ±2,00% |

Cuadruplicar el tamaño muestral reduce el margen de error a la mitad. El error disminuye con la raíz cuadrada de n, no linealmente.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# ── configuración ──────────────────────────────────────────────
p_hat   = 0.52        # proporción muestral
n       = 1200        # tamaño de la muestra
niveles = {
    "90% (α = 0,10)": {"alpha": 0.10, "color_zona": "#FAEEDA", "color_borde": "#BA7517", "color_cola": "#E24B4A"},
    "95% (α = 0,05)": {"alpha": 0.05, "color_zona": "#DAEAF7", "color_borde": "#185FA5", "color_cola": "#E24B4A"},
    "99% (α = 0,01)": {"alpha": 0.01, "color_zona": "#D4EFE6", "color_borde": "#0F6E56", "color_cola": "#E24B4A"},
}

# ── cálculos ───────────────────────────────────────────────────
def calcular_ic(p, n, alpha):
    z    = stats.norm.ppf(1 - alpha / 2)
    se   = np.sqrt(p * (1 - p) / n)
    me   = z * se
    return z, se, me, p - me, p + me

# ── figura ─────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle(
    "Intervalos de confianza — aprobación presidencial\n"
    r"$\hat{p}$ = 52%,  n = 1.200 ciudadanos argentinos",
    fontsize=13, fontweight="bold", y=1.02
)

x = np.linspace(-4, 4, 800)
y = stats.norm.pdf(x)

for ax, (titulo, cfg) in zip(axes, niveles.items()):
    z, se, me, lo, hi = calcular_ic(p_hat, n, cfg["alpha"])

    # ── curva completa (fondo gris claro) ──
    ax.plot(x, y, color=cfg["color_borde"], linewidth=1.8)

    # ── zona de confianza (central) ──
    mask_centro = (x >= -z) & (x <= z)
    ax.fill_between(x, y, where=mask_centro,
                    color=cfg["color_zona"], alpha=0.9,
                    label=f"Confianza: {int((1-cfg['alpha'])*100)}%")

    # ── colas de rechazo ──
    mask_cola = (x < -z) | (x > z)
    ax.fill_between(x, y, where=mask_cola,
                    color=cfg["color_cola"], alpha=0.35,
                    label=f"Rechazo: α = {cfg['alpha']}")

    # ── líneas z crítico ──
    for zv, lbl in [(-z, f"-z* = -{z:.3f}"), (z, f"+z* = {z:.3f}")]:
        ax.axvline(zv, color=cfg["color_borde"],
                   linewidth=1.2, linestyle="--", alpha=0.8)
        ax.text(zv, stats.norm.pdf(zv) + 0.005, lbl,
                ha="center", va="bottom", fontsize=7.5,
                color=cfg["color_borde"])

    # ── anotación del intervalo ──
    ax.text(0, -0.055,
            f"IC = [{lo*100:.2f}% ; {hi*100:.2f}%]\n"
            f"Margen de error = ±{me*100:.2f}%",
            ha="center", va="top", fontsize=8.5,
            color="#444441",
            bbox=dict(boxstyle="round,pad=0.4",
                      facecolor="white", edgecolor="#cccccc",
                      linewidth=0.8))

    # ── estética ──
    ax.set_title(titulo, fontsize=10, fontweight="bold",
                 color=cfg["color_borde"], pad=10)
    ax.set_xlabel("Valor z", fontsize=9)
    ax.set_ylabel("Densidad", fontsize=9) if ax == axes[0] else None
    ax.set_xlim(-4, 4)
    ax.set_ylim(-0.11, 0.45)
    ax.spines[["top", "right"]].set_visible(False)
    ax.tick_params(labelsize=8)
    ax.legend(fontsize=7.5, loc="upper right",
              framealpha=0.7, edgecolor="#cccccc")
    ax.set_xticks([-z, -2, -1, 0, 1, 2, z])
    ax.set_xticklabels(
        [f"{-z:.2f}", "-2", "-1", "0", "1", "2", f"{z:.2f}"],
        fontsize=7.5
    )

plt.tight_layout()
plt.savefig("intervalos_confianza.png", dpi=150,
            bbox_inches="tight", facecolor="white")
plt.show()
print("Archivo guardado: intervalos_confianza.png")
```

## Distribución muestral

Es una distribución teórica de probabilidades de los valores posibles de una variable aleatoria, que ocurriría si tuviesemos muestras posibles de un tamaño fijo de población

### Distribución normal: 
capacidad para modelar comportamientos donde la mayoría de las concentraciones se concentran alrededor de un valor central, mientras que los valores extremos son menos frecuentes.

## Distribución muestral: estandarización
Frecuentemente se estandariza las distribuciones normales para llevarlas a un normal estandar o distribución t de student. Este procedimiento consiste en tomar la variable que se considera que sigue una distribución normal y *restarle su media y dividirla por su desviación estándar*


## Estadísticos: prueba de hipótesis
Estos estadísticos se usan para saber o determinar si la evidencia en datos es suficiente para rechazar la hipótesis nula $H_0$ en favor de la alternativa $H_1$.

### Prueba Z
La prueba Z se utiliza generalmente cuando se conoce la Desviación Estándar (DV) de la población o cuando el tamaño de la muestra es lo suficientemente grande como para que la estimación de la desviación estándar se muestre confiable. Se basa en la distribución normal estándar y se utiliza para hacer desviaciones sobre la media poblacional *cuando los datos son aproximadamente normales* o el tamaño de la muestra es grande. 

### Prueba t de student
La prueba t de Student se utiliza cuando no se conoce la Desviación Estándar (DV) de la población y se debe estimar a partir de una muestra. También se utiliza cuando *el tamaño de una muestra es pequeño* y la distribución de los datos no es necesariamente normal. la distribución t tiene las colas mas gruesas que la distribución normal, lo que **refleja mayor incertidumbre** en la estimación del parámetro.

### Comparación entre estadístico y valor crítico
La regla general es comparar el valor absoluto del estadístico de prueba con el valor absoluto correspondiente al nivel de significación $\alpha$ previamente establecido.
- Si el estadístico de prueba calculado supera el valor crítico del valor absoluto (es decir, si cae en la *región de rechazo*), se rechaza la hpótesis nula.
- Si el estadístico de prueba calculado no supera el valor crítico (si cae dentro de la *región de rechazo*), no se rechaza la hipótesis nula.

***Es importante destacar que no rechazar la hpótesis nula no significa que la $H_0$ sea verdadera, sino que significa que no tiene suficiente evidencia para rechazarla con el nivel de significación elegido***

## 1. ¿Por qué estandarizar?

Las variables políticas se miden en escalas distintas: porcentajes de votos, índices de confianza de 0 a 10, cantidad de escaños, ingresos en pesos. Estandarizar convierte cualquier valor en una unidad común, sea el **valor z o valor t**, que expresa *cuántos desvíos estándar se aleja un resultado de la media*. Eso permite comparar resultados de distintas variables y ubicarlos en la distribución normal para hacer inferencia.

## 2. Estandarización con distribución normal (z)

Se usa cuando:
- el desvío estándar **poblacional (σ) es conocido**, o
- la muestra es **grande (n ≥ 30)**

### Fórmula para una observación

$$z = \frac{x - \mu}{\sigma}$$

### Fórmula para una media muestral

$$z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$$

### Fórmula para una proporción muestral

$$z = \frac{\hat{p} - \pi_0}{\sqrt{\dfrac{\pi_0(1-\pi_0)}{n}}}$$

### 3. Estandarización con distribución t de Student

Se usa cuando:
- el desvío estándar **poblacional (σ) es desconocido**
- se lo reemplaza por el **desvío muestral (s)**
- la muestra es **pequeña (n < 30)**

La distribución t es más achatada que la normal estándar y varía según los **grados de libertad (gl = n − 1)**. A medida que n crece, la t se aproxima a la normal.

### Fórmula para una media muestral

$$t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}$$

### Fórmula para diferencia de medias (dos grupos)

$$t = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\dfrac{s_1^2}{n_1} + \dfrac{s_2^2}{n_2}}}$$

Donde:
- $\bar{x}$ = media muestral
- $\mu_0$ = valor del parámetro bajo H₀
- $s$ = desvío estándar muestral
- $n$ = tamaño de la muestra
- $gl = n - 1$ grados de libertad

## Ejemlos: Estandarización y distribución t de Student

### 1. Ejemplo 1: estandarización z (n grande)

**Contexto:** Se encuestan n = 1.200 ciudadanos argentinos sobre confianza institucional (escala 0–10). De estudios anteriores se conoce σ = 2,1 (poblacional). La muestra da x̄ = 4,3. ¿Es significativamente distinto de 5?

$$H_0: \mu = 5 \qquad H_1: \mu \neq 5 \qquad \alpha = 0{,}05$$

**Estandarización:**

$$z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}} = \frac{4{,}3 - 5}{2{,}1 / \sqrt{1200}} = \frac{-0{,}7}{2{,}1 / 34{,}64} = \frac{-0{,}7}{0{,}0606} \approx -11{,}55$$

**Valor crítico** para prueba bilateral con α = 0,05:

$$z^* = \pm 1{,}96$$

**Decisión:**

$$|z| = 11{,}55 > 1{,}96 \quad \Rightarrow \quad \text{Se rechaza } H_0$$

**Interpretación:** La confianza institucional promedio de los argentinos es significativamente distinta de 5 puntos. Con una media muestral de 4,3, el resultado se ubica 11,55 desvíos estándar por debajo del valor de referencia: una diferencia estadísticamente muy significativa.

### 2. Ejemplo 2: estandarización t (n pequeño)

**Contexto:** Se entrevistan n = 18 concejales de municipios bonaerenses sobre su índice de acuerdo con el oficialismo
provincial (escala 0–10). La muestra da x̄ = 6,2 y s = 1,8. ¿Es el acuerdo significativamente mayor que 5?

$$H_0: \mu \leq 5 \qquad H_1: \mu > 5 \qquad \alpha = 0{,}05$$

**Grados de libertad:**

$$gl = n - 1 = 18 - 1 = 17$$

**Estandarización:**

$$t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}} = \frac{6{,}2 - 5}{1{,}8 / \sqrt{18}} = \frac{1{,}2}{1{,}8 / 4{,}243} = \frac{1{,}2}{0{,}4243} \approx 2{,}828$$

**Valor crítico** para prueba de cola derecha,
α = 0,05, gl = 17:

$$t^*_{(0{,}05;\,17)} = 1{,}740$$

**Decisión:**

$$t = 2{,}828 > 1{,}740 \quad \Rightarrow \quad \text{Se rechaza } H_0$$

**Interpretación:** El índice de acuerdo de los concejales con el oficialismo provincial es significativamente mayor
que 5, el punto medio de la escala. Hay evidencia de alineamiento político con el gobierno provincial.

### 3. Ejemplo 3 — diferencia de medias t (dos grupos)

**Contexto:** Se compara el índice de confianza institucional entre ciudadanos de municipios con presupuesto participativo (Grupo 1, n₁ = 12) y sin presupuesto participativo (Grupo 2, n₂ = 14). ¿Difieren significativamente?

| | Grupo 1 (con PP) | Grupo 2 (sin PP) |
|:--|:--:|:--:|
| n | 12 | 14 |
| x̄ | 6,1 | 4,8 |
| s | 1,5 | 1,9 |

$$H_0: \mu_1 = \mu_2 \qquad H_1: \mu_1 \neq \mu_2 \qquad \alpha = 0{,}05$$

**Estandarización:**

$$t = \frac{6{,}1 - 4{,}8}{\sqrt{\dfrac{1{,}5^2}{12} + \dfrac{1{,}9^2}{14}}} = \frac{1{,}3}{\sqrt{\dfrac{2{,}25}{12} + \dfrac{3{,}61}{14}}} = \frac{1{,}3}{\sqrt{0{,}1875 + 0{,}2579}} = \frac{1{,}3}{\sqrt{0{,}4454}} = \frac{1{,}3}{0{,}6674} \approx 1{,}948$$

**Grados de libertad aproximados (Welch):**

$$gl \approx \frac{(0{,}1875 + 0{,}2579)^2}{\dfrac{0{,}1875^2}{11} + \dfrac{0{,}2579^2}{13}} = \frac{0{,}1985}{0{,}003196 + 0{,}005117} \approx \frac{0{,}1985}{0{,}008313} \approx 23$$

**Valor crítico** bilateral, α = 0,05, gl = 23:

$$t^*_{(0{,}025;\,23)} = \pm 2{,}069$$

**Decisión:**

$$|t| = 1{,}948 < 2{,}069 \quad \Rightarrow \quad \text{No se rechaza } H_0$$

**Interpretación:** No hay evidencia estadística suficiente para afirmar que el presupuesto participativo produce diferencias significativas en la confianza institucional con este tamaño muestral. Se requeriría una muestra mayor.

## Errores en la inferencia estadística
- Error tipo 1 (Falso Positivo): El investigador concluye erróneamente que existe una diferencia significativa o un efecto cuando en realidad no lo  hay.

Ejemplo: Afirmar que un programa de asistencia social reduce la pobreza cuando, en realidad, el cambio observado se debió a factores externos y el programa no tuvo ninún efecto.

$\alpha$ se define *a priori* siendo 0.5 y 0.1 valores habituales y es asimilable al "margen de error" que estoy dsipuesto a aceptar.

- Error tipo 2 (Falso negativo): El investigador falla al detectar una evidencia o efecto significativo que sí existe en la población.

Ejemplo: Determinar que una nueva ley de seguridad no ha tenido ningpun impacto en la reducción de la criminalidad, cuando en realidad sí funcionó pero no se contó con una muestra *lo suficientemente grande* para detectarlo.

$\beta$ depende del tamaño de la muestra

### Regla de decisión:
Se fija un valor de $\alpha$ como máximo error de decisión tolerado.
Se calcula la probabilidad de equivocarse al rechazar $H_0$, con los datos de la muestra (Valor-p o *p-value*)

- Rechazo $H_0$ si el *p-value* es menor o igual que $\alpha$
- No rechazo $H_0$ si el *p-value* no es menor que $\alpha$

## Tests paramétricos

### Prueba de hipótesis para una media frente a un parámetro
Test de t para dos medias independientes.
 $X_11, ..., X_1n$ muestra de una población normal con media $\mu_1$ y desvío $\sigma_1$
 $X_21, ..., X_2m$ muestra de una población normal con media $\mu_2$ y desvío $\sigma_2$

 $$ H_0: \mu_1 - \mu_2 = \delta_0 $$

 $$ H_\alpha: \mu_1 - \mu_2 = \neq \delta_0
 H_\alpha: \mu_1 - \mu_2 = > \delta_0
 H_\alpha: \mu_1 - \mu_2 = < \delta_0 $$
 Parámtros habituales para definir que tipo de prueba estoy realizando de paquete estadístico en Python y R.

$H_0 y H_\alpha$ son eventos exhaustivos y mutuamente excluyentes. Una de las dos afirmaciones *debe* ser verdadera.

### Supuestos de la pruebta t para comparación de dos medias independientes
- Independencia: Las observaciones tienen que ser independientes unas de las otras.
- Normalidad: Las poblaciones que se comparan tiene que tener una distribución normal. En caso de asimetría, los t-test son robustos si el tamaño de las muestras es mayor o igual a 30.
- Igualdad de varianza (homocelasticidad): La varianza de las poblaciones comparadas debe ser igual. En caso de no cumplirse se puede emplerar un **Welch Two Sample t-test**

*Es importante evaluar hasta qué punto el no cumplimiento de una o de varias de sus condiciones puede afectar el resultado del t-test. Si la utilización queda descartada, se puede recurrir a otros tests estadísticos*

### Test de t para dos medias APAREADAS
Se considera que la variable $d_i = X_1i - X_2i$ para cada uno de los sujetos de la muestra.
con esto, puede comprobarse que:

$ \bar{d} = \bar{X}_1 - \bar{X}_2$

El estadístico de prueba es 

$$E = \frac{\bar{X}_1 - \bar{X}_2 - \Delta_0} {S_d / \sqrt{n}} \sim t_{n-1}$$


### Supuestos:
- Normalidad en cada una de las poblaciones

Problema: analizar si el escándalo de la promoción de LIBRA disminuyó la intensisdad de la confianza en Javier Milei

Medición: se encuesta a las mismas 500 personas en dos momentos:
- 1, antes: se les pregunta: *en una escala del 1 al 10, donde 1 es 'ninguna confianza' y 10 'absoluta cofianza', ¿qué tan confiable le parece Javier Milei?
- 2, después: tras el escándalo, se vuelve a hacer exactamente la misma pregunta.

Hipótesis (prueba t con muestras apareadas): 
- $H_0$ (nula): El promedio de confianza de Milei antes del escándalo *es igual* al promedio de confianza después. donde: $\mu_a = \mu_d$
- $H_a$ (alternativa): El promedio de confianza de milei es *diferente* después del escándalo. Donde: $\mu_d < \mu_a$

### Tests no paramétricos

#### Test de Mann-Withney para comparación de dos medias
- Es un test no paramétrico que demuestra si dos poblaciones provienen de muestras equidistribuidas
- La idea en la que se fundamenta es la siguiente: Si las dos muestras comparadas proceden de la misma población, al juntar todas las observaciones y ordenarlas de menor a mayor, cabría esperar que las observaciones de una y otra muestra estuviesen intercaladas aleatoriamente.
Por lo contrario, si una población pertenece a una muestra con mayores o menores valores que la otra población, al ordenar las observaciones, éstas tenderan a agruparse de modo que las de una muestra queden por encima de la otra.

Se basa en la combinación de los conjuntos $N_1$ y $N_2$ observaciones
Cuando todas las observaciones están juntas se les asigna un rango a cada una, *una vez ordenadas*, que comienza en 1 y termina en $N_1 + N_2$

Procedimiento:
Se obtiene la suma de los rangos asociados a las observaciones de una de las muestras. Esa suma es = a $R_1$.

El Estadístico está dado por:

$$U = n_1n_2 + \frac{n_1(n_1 + 1)}{2} - R_1$$

Condiciones necesarias: 
- Independencia: Las observaciones tienen que ser independientes unas de las otras. Para ello, el muestreo debe ser aleatorio.
- La variable tiene que ser *por lo menos ordinal*, o bien deben poder ordenarse de menor a mayor.
- No es necesario asumir que las muestras se distribuyen de forma normal o que proceden de poblaciones normales. Pero, para que el test compare medias, ambas han de tener el mismo tipo de distribución.
- Igualdad de varianza entre grupos. **(homocelasticidad)**

## Ejemplos: Tests paramétricos y no paramétricos en Ciencia Política

### 1. Prueba de hipótesis para una media frente a un parámetro

Se usa cuando se quiere comparar la media de una muestra contra un valor de referencia teórico o histórico conocido.

### Supuestos

- Variable cuantitativa continua
- Distribución aproximadamente normal (o n ≥ 30)
- σ desconocido → se usa t de Student
- Una sola muestra

### Fórmula

$$t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}} \qquad gl = n - 1$$

**Pregunta:** ¿El promedio de confianza institucional de los ciudadanos de municipios con presupuesto participativo es distinto del promedio nacional histórico de 4,5 puntos?

Se encuestan n = 25 ciudadanos de municipios con PP.

$$H_0: \mu = 4{,}5 \qquad H_1: \mu \neq 4{,}5 \qquad \alpha = 0{,}05$$

**Datos muestrales:**

$$\bar{x} = 5{,}8 \qquad s = 1{,}6 \qquad n = 25$$

**Cálculo del estadístico:**

$$t = \frac{5{,}8 - 4{,}5}{1{,}6 / \sqrt{25}} = \frac{1{,}3}{1{,}6 / 5} = \frac{1{,}3}{0{,}32} = 4{,}063$$

**Valor crítico:** $t^*_{(0{,}025;\,24)} = 2{,}064$

**Decisión:**

$$|t| = 4{,}063 > 2{,}064 \quad \Rightarrow \quad \text{Se rechaza } H_0$$

**Interpretación:** Los ciudadanos de municipios con presupuesto participativo presentan una confianza institucional significativamente mayor que el promedio nacional histórico de 4,5 puntos.

## 2. Prueba t para dos medias independientes

Se usa cuando se comparan las medias de **dos grupos distintos y sin relación entre sí**. Cada unidad pertenece a uno solo de los dos grupos.

### Supuestos

- Variable cuantitativa continua
- Los dos grupos son **independientes** entre sí
- Distribución aproximadamente normal en cada grupo
- σ desconocido en ambos grupos
- Se evalúa si las varianzas son iguales o distintas
  (test de Levene) para elegir la versión correcta

### Evaluación de varianzas: test de Levene

Antes de aplicar la t para dos medias independientes, se verifica si las varianzas de ambos grupos son iguales:

$$H_0: \sigma_1^2 = \sigma_2^2 \qquad H_1: \sigma_1^2 \neq \sigma_2^2$$

- Si **no se rechaza H₀** (varianzas iguales):
  se usa la t con varianzas agrupadas (pooled)
- Si **se rechaza H₀** (varianzas distintas):
  se usa la t de Welch

### Fórmula con varianzas iguales (pooled)

$$t = \frac{\bar{x}_1 - \bar{x}_2}{s_p \sqrt{\dfrac{1}{n_1} + \dfrac{1}{n_2}}}$$

$$s_p = \sqrt{\frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1+n_2-2}}
\qquad gl = n_1 + n_2 - 2$$

### Fórmula con varianzas distintas (Welch)

$$t = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\dfrac{s_1^2}{n_1} + \dfrac{s_2^2}{n_2}}}
\qquad gl \approx \frac{\left(\dfrac{s_1^2}{n_1}+\dfrac{s_2^2}{n_2}\right)^2}
{\dfrac{(s_1^2/n_1)^2}{n_1-1}+\dfrac{(s_2^2/n_2)^2}{n_2-1}}$$

**Pregunta:** ¿Los legisladores de partidos de izquierda tienen un índice de actividad parlamentaria
(proyectos presentados por año) distinto al de los legisladores de partidos de derecha?

$$H_0: \mu_1 = \mu_2 \qquad H_1: \mu_1 \neq \mu_2 \qquad \alpha = 0{,}05$$

**Datos:**

| | Izquierda (G1) | Derecha (G2) |
|:--|:--:|:--:|
| n | 18 | 20 |
| x̄ | 14,3 proyectos | 10,8 proyectos |
| s | 3,2 | 4,1 |

**Test de Levene:** p = 0,18 → no se rechaza H₀ → varianzas iguales → se usa pooled.

**Varianza agrupada:**

$$s_p = \sqrt{\frac{17 \times 3{,}2^2 + 19 \times 4{,}1^2}{18+20-2}}
= \sqrt{\frac{17 \times 10{,}24 + 19 \times 16{,}81}{36}}
= \sqrt{\frac{174{,}08 + 319{,}39}{36}}
= \sqrt{\frac{493{,}47}{36}} = \sqrt{13{,}71} \approx 3{,}70$$

**Estadístico:**

$$t = \frac{14{,}3 - 10{,}8}{3{,}70\sqrt{\dfrac{1}{18}+\dfrac{1}{20}}}
= \frac{3{,}5}{3{,}70 \times 0{,}320} = \frac{3{,}5}{1{,}184} \approx 2{,}956$$

**Valor crítico:** $t^*_{(0{,}025;\,36)} \approx 2{,}028$

**Decisión:**

$$|t| = 2{,}956 > 2{,}028 \quad \Rightarrow \quad \text{Se rechaza } H_0$$

**Interpretación:** Los legisladores de izquierda presentan significativamente más proyectos por año que los de derecha. La diferencia de 3,5 proyectos promedio es estadísticamente significativa con α = 0,05.

### 3. Test t para dos medias apareadas

Se usa cuando **las mismas unidades** son medidas en dos momentos distintos o bajo dos condiciones distintas. No son grupos independientes: cada observación del grupo 1 tiene su par en el grupo 2.

### Supuestos

- Variable cuantitativa continua
- **Las observaciones están emparejadas** (misma unidad, dos mediciones)
- Las diferencias dᵢ = x₁ᵢ − x₂ᵢ se distribuyen normalmente
- Se trabaja con las diferencias, no con los valores originales

### Fórmula

$$d_i = x_{1i} - x_{2i}$$

$$\bar{d} = \frac{\sum d_i}{n} \qquad
s_d = \sqrt{\frac{\sum(d_i - \bar{d})^2}{n-1}}$$

$$t = \frac{\bar{d}}{s_d / \sqrt{n}} \qquad gl = n - 1$$

**Pregunta:** ¿La aprobación presidencial cambió significativamente antes y después de una cadena nacional de anuncios económicos?

Se mide la aprobación (escala 0–10) de los **mismos**
n = 12 grupos focales, antes y después del anuncio.

$$H_0: \mu_d = 0 \qquad H_1: \mu_d \neq 0 \qquad \alpha = 0{,}05$$

| Grupo | Antes (x₁) | Después (x₂) | dᵢ = x₁ − x₂ |
|:-----:|:----------:|:------------:|:-------------:|
| 1     | 5,2        | 6,1          | −0,9          |
| 2     | 4,8        | 5,9          | −1,1          |
| 3     | 6,1        | 6,8          | −0,7          |
| 4     | 3,9        | 5,2          | −1,3          |
| 5     | 5,5        | 6,3          | −0,8          |
| 6     | 4,2        | 4,9          | −0,7          |
| 7     | 6,3        | 7,1          | −0,8          |
| 8     | 5,0        | 5,8          | −0,8          |
| 9     | 4,7        | 5,5          | −0,8          |
| 10    | 5,8        | 6,6          | −0,8          |
| 11    | 3,5        | 4,8          | −1,3          |
| 12    | 6,0        | 6,9          | −0,9          |

$$\bar{d} = \frac{-9{,}9}{12} = -0{,}825$$

$$s_d \approx 0{,}196$$

**Estadístico:**

$$t = \frac{-0{,}825}{0{,}196 / \sqrt{12}} = \frac{-0{,}825}{0{,}0566} \approx -14{,}57$$

**Valor crítico:** $t^*_{(0{,}025;\,11)} = \pm 2{,}201$

**Decisión:**

$$|t| = 14{,}57 > 2{,}201 \quad \Rightarrow \quad \text{Se rechaza } H_0$$

**Interpretación:** La cadena nacional produjo un aumento significativo en la aprobación presidencial. En promedio, la aprobación subió 0,825 puntos en todos los grupos focales, y esa diferencia es estadísticamente muy significativa.

### 4. Tests no paramétricos

Los tests no paramétricos se usan cuando **no se cumplen los supuestos de normalidad** de los tests paramétricos, especialmente con muestras pequeñas o variables ordinales.

No asumen una distribución específica en los datos. En lugar de trabajar con los valores originales, trabajan con los **rangos** (posiciones ordenadas).

| Situación paramétrica | Alternativa no paramétrica |
|:--|:--|
| t para una media | Test de Wilcoxon de los rangos con signo |
| t para dos medias independientes | Test de Mann-Whitney (U) |
| t para medias apareadas | Test de Wilcoxon apareado |

### 4a. Test de Mann-Whitney (U)

**Equivalente no paramétrico de la t para dos medias independientes.**
No requiere normalidad. Compara si una distribución tiende a tener valores más altos que la otra.

### Hipótesis

$$H_0: \text{las dos distribuciones son iguales}$$
$$H_1: \text{una distribución tiende a valores mayores}$$

### Procedimiento

1. Se combinan todos los datos de ambos grupos y se ordenan
2. Se asigna un rango a cada valor (1 = menor)
3. Se suman los rangos de cada grupo por separado
4. Se calcula el estadístico U

### Fórmula

$$U_1 = n_1 n_2 + \frac{n_1(n_1+1)}{2} - R_1$$

$$U_2 = n_1 n_2 + \frac{n_2(n_2+1)}{2} - R_2$$

$$U = \min(U_1, U_2)$$

Para muestras grandes (n > 20) se aproxima con z:

$$z = \frac{U - \dfrac{n_1 n_2}{2}}{\sqrt{\dfrac{n_1 n_2 (n_1+n_2+1)}{12}}}$$

**Pregunta:** ¿Los concejales de municipios rurales tienen distinto índice de gestión de recursos (0–100) que los de municipios urbanos?

Los datos **no cumplen normalidad** (Shapiro-Wilk rechazado) → se usa Mann-Whitney.

$$H_0: \text{distribuciones iguales} \qquad H_1: \text{distribuciones distintas} \qquad \alpha = 0{,}05$$

**Datos:**

| 55 | 73 |
| 38 | 58 |
| 47 | 69 |
| 51 | 65 |
| 44 | 72 |

**Paso 2 — sumar rangos por grupo:**

$$R_1 = 1+2+3+4+5+6 = 21 \qquad R_2 = 7+8+9+10+11+12 = 57$$

**Paso 3 — calcular U:**

$$U_1 = 6 \times 6 + \frac{6 \times 7}{2} - 21 = 36 + 21 - 21 = 36$$

$$U_2 = 6 \times 6 + \frac{6 \times 7}{2} - 57 = 36 + 21 - 57 = 0$$

$$U = \min(36, 0) = 0$$

**Valor crítico** para n₁ = n₂ = 6, α = 0,05 bilateral: $U^* = 5$

**Decisión:**

$$U = 0 < 5 \quad \Rightarrow \quad \text{Se rechaza } H_0$$

**Interpretación:** Los concejales de municipios urbanos tienen índices de gestión significativamente más altos que los rurales. La separación total de rangos (todos los rurales con rangos bajos, todos los urbanos con rangos altos) indica una diferencia muy marcada.

### 4b. Test de Wilcoxon para medias apareadas

**Equivalente no paramétrico de la t apareada.**
Se usa cuando los datos están emparejados pero las diferencias no se distribuyen normalmente.

### Hipótesis

$$H_0: \text{la mediana de las diferencias es cero}$$
$$H_1: \text{la mediana de las diferencias no es cero}$$

### Procedimiento

1. Calcular las diferencias dᵢ = x₁ᵢ − x₂ᵢ
2. Eliminar los empates (dᵢ = 0)
3. Ordenar los |dᵢ| de menor a mayor y asignar rangos
4. Separar rangos según el signo de dᵢ
5. Calcular W⁺ (suma de rangos positivos) y W⁻ (negativos)
6. El estadístico es W = min(W⁺, W⁻)

Para n > 25 se aproxima con z:

$$z = \frac{W - \dfrac{n(n+1)}{4}}{\sqrt{\dfrac{n(n+1)(2n+1)}{24}}}$$

### Ejemplo

**Pregunta:** ¿El índice de polarización afectiva de los mismos n = 10 distritos electorales cambió significativamente entre 2019 y 2023?

Las diferencias **no cumplen normalidad** → se usa Wilcoxon.

$$H_0: \text{mediana}_d = 0 \qquad H_1: \text{mediana}_d \neq 0 \qquad \alpha = 0{,}05$$

| Distrito | 2019 (x₁) | 2023 (x₂) | dᵢ   | \|dᵢ\| | Rango | Signo |
|:--------:|:---------:|:---------:|:----:|:------:|:-----:|:-----:|
| 1        | 5,2       | 6,8       | −1,6 | 1,6    | 7     | −     |
| 2        | 4,8       | 5,9       | −1,1 | 1,1    | 4     | −     |
| 3        | 6,1       | 7,4       | −1,3 | 1,3    | 5,5   | −     |
| 4        | 3,9       | 3,4       | +0,5 | 0,5    | 1     | +     |
| 5        | 5,5       | 6,8       | −1,3 | 1,3    | 5,5   | −     |
| 6        | 4,2       | 4,9       | −0,7 | 0,7    | 2     | −     |
| 7        | 6,3       | 7,4       | −1,1 | 1,1    | 4     | −     |
| 8        | 5,0       | 5,8       | −0,8 | 0,8    | 3     | −     |
| 9        | 4,7       | 6,2       | −1,5 | 1,5    | 6     | −     |
| 10       | 5,8       | 7,6       | −1,8 | 1,8    | 8     | −     |

> Los rangos 5 y 6 comparten el mismo \|d\| = 1,3
> → se asigna el rango promedio: (5+6)/2 = 5,5

**Suma de rangos:**

$$W^+ = 1 \qquad W^- = 7+4+5{,}5+5{,}5+2+4+3+6+8 = 45$$

$$W = \min(1, 45) = 1$$

**Valor crítico** para n = 10, α = 0,05 bilateral: $W^* = 8$

**Interpretación:** La polarización afectiva aumentó significativamente entre 2019 y 2023 en los distritos estudiados. Solo 1 de los 10 distritos mostró una reducción (Distrito 4), mientras los 9 restantes registraron aumentos, lo que produce una suma de rangos negativos muy dominante.

## 5. Resumen comparativo

| Test | Tipo | Supuesto clave | Ejemplo político |
|:--|:--:|:--|:--|
| t una media | Paramétrico | Normalidad, σ desconocido | Confianza inst. vs histórico |
| t dos medias independientes | Paramétrico | Normalidad, grupos distintos | Proyectos izq. vs der. |
| t apareada | Paramétrico | Normalidad de diferencias | Aprobación antes/después |
| Mann-Whitney | No paramétrico | Sin normalidad, grupos distintos | Gestión rural vs urbano |
| Wilcoxon apareado | No paramétrico | Sin normalidad, datos apareados | Polarización 2019 vs 2023 |

> **Regla práctica:** si la muestra es pequeña (n < 30)
> y el test de Shapiro-Wilk rechaza la normalidad,
> usar el equivalente no paramétrico.
> Si n ≥ 30, el teorema central del límite permite
> usar tests paramétricos aunque los datos no sean normales.

## Conclusiones
### ¿Prueba paramétrica o No Paramétrica?
**Use un método paramétrico si...**
- Sus datos son numéricos
- Se cumple el supuesto de normalidad en la distribución de los datos
- Si la muestra es lo ***suficientemente grande*** puede usarlo aunque los datos no sean suficientemente normales, gracias al Teorema del Límite Central

**Use un método No Paramétrico si...**
- Sus datos son ordinales
- Sus datos numéricos ***NO*** siguen una distribución normal
- Su muestra es muy pequeña y no tiene certeza que la población de origen sea normal.
